### Creating Arrays With a Defined Data Type

We use the `array()` function to create arrays, this function can take an optional argument: dtype that allows us to define the expected data type of the array elements.

```python
arr1 = np.array([1, 2, 3, 4], dtype='U') # we are defining the array as Unicode String with `U`. So the elements are becomes string.
print(arr1)
print(arr1.dtype)

arr2 = np.array([1, 2, 3, 4], dtype=np.int32) # we are defining the array as integer 32 with `int32`. So the elements are becomes integer 32.
print(arr2)
print(arr2.dtype)


arr3 = np.array(['1', '2', '3', '4'], dtype=np.uint64) # we are defining the array as unsigned integer 64 with `uint64`. So the elements are becomes unsigned integer 64.
print(arr3)
print(arr3.dtype)

arr4 = arr3.astype(np.float128) # we are converting an array data type
print(arr4)
print(arr4.dtype)
```

### Upcasting (Implicit & Automatic)

Upcasting happens when NumPy promotes a data type to a higher level of precision or a more general type so that no data is lost.

• When it occurs: Automatically (implicitly) when we mix different types in an array or perform operations between different dtypes (such as adding an integer array to a float array).

• Hierarchy example: bool -> int -> float -> complex.


```python
arr = np.array([1,2,3.2])
# upcasted to float
print(arr.dtype)

#upcasted to string
arr = np.array([1,2,3.2,'hello'])
print(arr.dtype)


arr = np.array([1,True,3.25,'hello',])
print(arr.dtype)
```
