```python
import numpy as np

single_random_value = np.random.rand() # it will return a single random value which will be floating value from 0 to 1.
print(single_random_value)
print(type(single_random_value))
```

```python
# np.random.rand(shape)

arr = np.random.rand(2,3) # it will return 2 by 3 array with random values which will be floating value from 0 to 1.

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# np.random.randint(start,end, shape)

arr = np.random.randint(1, 10, (1, 5)) # it will return 1 by 5 array with random values which will be integer value from 1 to 10.

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# np.random.randint(start,end, shape)

arr = np.random.randint(1,10 , (2,3)) # it will return 2 by 3 array with random values which will be integer value from 1 to 10.

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# np.random.uniform(start,end, shape)

arr = np.random.uniform(50,100, (3,3,4)) # it will return 3 by 3 by 4 array with random values which will be floating value from 50 to 100.

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```
