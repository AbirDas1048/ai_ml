### Transpose NumPy Array

To transpose a NumPy array, use the `.T` attribute or the `np.transpose()` function, both of which reverse or permute the array's axes. For standard 2D arrays, this swaps rows and columns while returning a view of the original data rather than a copy.

```python
import numpy as np

arr = np.random.randint(1 ,10 , size =(2,3))

print(arr, end="\n\n")

# Method 1: Using the .T attribute (most common)
print(arr.T, end="\n\n")

# Method 2: Using np.transpose()
print(np.transpose(arr), end="\n\n")
```
