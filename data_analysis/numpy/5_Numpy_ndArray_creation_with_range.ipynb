### Array creation with range functions

```python
import numpy as np
# np.arange(first , end , step ) -> similar to range function

arr = np.arange(1 , 10 ,1 )

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
mat = arr.reshape(3,3)

print(mat)
print(type(mat))
print(mat.dtype)
print(mat.shape)
print(mat.ndim)
print(mat.size)
```

### linspace

Generates an array of evenly spaced numbers over a specified interval

```python
# np.linspace(start , end(included) , how much point)

arr = np.linspace(0, 4 ,6) # It will generates an array with 0 to 4 and with same interval
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)

arr = np.linspace(0, 10, 20)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

### logspace

It generates numbers that are evenly spaced on a logarithmic scale. Instead of choosing the actual starting and ending numbers directly, we provide the exponents for a given base (which defaults to base 10)

```python
# np.logspace(start , end(included) , how much point, base default 10 )

arr = np.logspace(0, 4, 6)

print(arr)

arr = np.logspace(0, 4, 5, base = 2)

print(arr)
```
