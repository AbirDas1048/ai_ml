### Encapsulation
Encapsulation is about protecting data inside a class.

It means keeping data (properties) and methods together in a class, while controlling how the data can be accessed from outside the class.

This prevents accidental changes to our data and hides the internal details of how your class works.

### Type of property and method accessing name

• Public (name): Anyone can read or change it.

• Protected (_name): "Please treat me as private unless you are a child class."

• Private (__name): Strictly locked down. If we try to call it from outside, Python pretends it doesn't even exist.


```python
class BankAccount:
    def __init__(self, owner, account_number, balance):
        self.owner = owner               # Public: Can be accessed anywhere
        self._account_number = account_number # Protected: For internal/subclass use
        self.__balance = balance         # Private: Hidden from outside access

    # --- PRIVATE METHOD ---
    def __validate_transaction(self, amount):
        """Internal check that cannot be called from outside the class."""
        return amount > 0

    # --- PROTECTED METHOD ---
    def _display_statement_header(self):
        """Intended for this class and its subclasses."""
        print(f"--- Statement for {self.owner} ---")

    # --- PUBLIC METHODS (Getters and Setters) ---
    def get_balance(self):
        """Public getter to safely view private balance."""
        return self.__balance

    def deposit(self, amount):
        """Public setter to safely modify private balance."""
        if self.__validate_transaction(amount): # Accessing private method internally
            self.__balance += amount
            print(f"Deposited ${amount}. New balance: ${self.__balance}")
        else:
            print("Invalid deposit amount!")


class SavingsAccount(BankAccount):
    def show_account_details(self):
        # Accessible: Public properties are open
        print(f"Owner: {self.owner}") 
        
        # Accessible: Subclasses can access protected properties/methods
        print(f"Account Number: {self._account_number}") 
        self._display_statement_header() 
        
        # ERROR: Un-commenting the line below will crash the program.
        # Subclasses CANNOT access private properties of the parent class.
        # print(f"Balance: {self.__balance}") 

# --- TESTING ENCAPSULATION ---

account = BankAccount("Alice", "123-456", 1000)

# 1. Public properties are easy to read/write
print(account.owner) # Output: Alice

# 2. Protected properties work, but you shouldn't access them directly like this
print(account._account_number) # Output: 123-456 (Python allows it, but it violates convention)

# 3. Private properties are strictly locked
# print(account.__balance) # AttributeError: 'BankAccount' object has no attribute '__balance'
# account.__validate_transaction(500) # AttributeError

# 4. Correct way to interact with private data (using public methods)
account.deposit(250) # Output: Deposited $250. New balance: $1250
print("Balance via Getter:", account.get_balance()) # Output: Balance via Getter: 1250
```
