### Inheritance
`Inheritance` allows us to define a class that inherits all the methods and properties from another class.

Parent class is the class being inherited from, also called base class.

Child class is the class that inherits from another class, also called derived class.

#### Single Inheritance

-> Create a parent class (Phone)

-> Create a child class which is only inherit the parent class (Camera Phone)

-> Create objects from parent class and child class both

```python
class Phone:
    def __init__(self, brand, model, battery_percerntage = 100):
        self.brand = brand
        self.model = model
        self.battery_percerntage = battery_percerntage

    def get_battery_percerntage(self):
        return self.battery_percerntage

    def show_battery_percerntage(self):
        print(f"{self.model} phone Battery percentage is {self.battery_percerntage}")

    def display_message(self, message):
        print(f"{self.model} phone {message}")

    def check_battery_percentage(self, for_charge = False, for_use = False):
        percentage = self.get_battery_percerntage()
        if(percentage <= 0):
            if(for_use):
                self.display_message("low battery please plugin to charge")
                return False
            else:
                return True
        elif(percentage >= 100):
            if(for_charge):
                self.display_message("battery is fully charged")
                return False
            else:
                return True
        else:
            return True

    def plugin_to_charge(self):
        if(self.check_battery_percentage(for_charge=True)):
            self.increase_battery_percerntage(5)
            self.show_battery_percerntage()
            
    def decress_battery_percerntage(self, battery_percerntage):
        self.battery_percerntage -= battery_percerntage

    def increase_battery_percerntage(self, battery_percerntage):
        self.battery_percerntage += battery_percerntage

    def call(self):
        if(self.check_battery_percentage(for_use=True)):
            self.display_message("calling normal phone")
            self.decress_battery_percerntage(5)

class CameraPhone(Phone):
    def __init__(self, brand, model, battery_percerntage = 100, camera = '2MP'):
        super().__init__(brand, model, battery_percerntage)
        self.camera = camera

    def listening_music(self):
        if(super().check_battery_percentage(for_use=True)):
            super().display_message("listening music")
            super().decress_battery_percerntage(5)

    def capturing_image(self):
        if(super().check_battery_percentage(for_use=True)):
            super().display_message("image captured")
            super().decress_battery_percerntage(10)

    def call(self):
        if(super().check_battery_percentage(for_use=True)):
            super().display_message("calling camera phone")
            super().decress_battery_percerntage(10)


nokia = Phone('Nokia', '1100')
nokia.show_battery_percerntage()
nokia.call()
# nokai.capturing_image() # will be show error

nokia_c101 = CameraPhone('Nokia', 'C101', camera='3.5MP')
nokia_c101.capturing_image()
nokia_c101.call()
nokia_c101.show_battery_percerntage()

```
