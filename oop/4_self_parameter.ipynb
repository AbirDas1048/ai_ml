### The `self` Parameter
The `self` parameter is a reference to the current instance of the class.

It is used to access properties and methods that belong to the class.

Why Use `self`?
Without `self`, Python would not know which object's properties we want to access.

Accessing Properties and Methods with `self`

```python
class Phone:
    def __init__(self, brand, model, battery_percerntage = 100):
        self.brand = brand
        self.model = model
        self.battery_percerntage = battery_percerntage

    def get_battery_percerntage(self):
        return self.battery_percerntage

    def show_battery_percerntage(self):
        print(f"{self.model} phone Battery percentage is {self.battery_percerntage}")

    def display_message(self, message):
        print(f"{self.model} phone {message}")

    def check_battery_percentage(self, for_charge = False, for_use = False):
        percentage = self.get_battery_percerntage()
        if(percentage <= 0):
            if(for_use):
                self.display_message("low battery please plugin to charge")
                return False
            else:
                return True
        elif(percentage >= 100):
            if(for_charge):
                self.display_message("battery is fully charged")
                return False
            else:
                return True
        else:
            return True

    def plugin_to_charge(self):
        if(self.check_battery_percentage(for_charge=True)):
            self.increase_battery_percerntage(5)
            self.show_battery_percerntage()
            
    def decress_battery_percerntage(self, battery_percerntage):
        self.battery_percerntage -= battery_percerntage

    def increase_battery_percerntage(self, battery_percerntage):
        self.battery_percerntage += battery_percerntage

    def listening_music(self):
        if(self.check_battery_percentage(for_use=True)):
            print(f"{self.brand} phone listening music")
            self.decress_battery_percerntage(5)
            
    def playing_games(self):
        if(self.check_battery_percentage(for_use=True)):
            print(f"{self.brand} phone playing games")
            self.decress_battery_percerntage(20)

    def capturing_image(self):
        if(self.check_battery_percentage(for_use=True)):
            print(f"{self.brand} phone image captured")
            self.decress_battery_percerntage(10)

apple = Phone("Apple", "iPhone 13")
samsung = Phone("Samsung", "Galaxy S21")

apple.show_battery_percerntage()
samsung.show_battery_percerntage()

apple.playing_games()
apple.listening_music()
apple.capturing_image()

samsung.playing_games()
samsung.listening_music()
samsung.listening_music()
samsung.listening_music()
samsung.listening_music()
samsung.playing_games()
samsung.capturing_image()

apple.show_battery_percerntage()
samsung.show_battery_percerntage()

apple.plugin_to_charge()
apple.plugin_to_charge()

samsung.playing_games()
samsung.capturing_image()
samsung.listening_music()

apple.show_battery_percerntage()
samsung.show_battery_percerntage()

samsung.plugin_to_charge()
samsung.plugin_to_charge()

apple.playing_games()
apple.capturing_image()
apple.listening_music()

apple.show_battery_percerntage()
samsung.show_battery_percerntage()
```

### `self` Does Not Have to Be Named "self"
It does not have to be named `self`, we can call it whatever we like, but it has to be the first parameter of any method in the class.

```python
# Use the words myobject and abc, xyz instead of self.

class Person:
    def __init__(myobject, name, age):
        myobject.name = name
        myobject.age = age

    def greet(abc):
        print("Hello, my name is " + abc.name)

    def show_age(xyz):
        print(f"My age is {xyz.age}")


p1 = Person("Abir", 32)
p1.greet()
p1.show_age()
```
