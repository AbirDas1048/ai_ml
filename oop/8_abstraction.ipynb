### Abstraction

Abstract class acts as a blueprint or contract. It cannot be instantiated directly, and it forces any child class to implement specific methods before it can be used.

Abstraction is the process of hiding complex implementation details and showing only the essential features of an object. We can achieve this by creating Abstract Base Classes (ABC).

```python
from abc import ABC, abstractmethod

# --- ABSTRACT CLASS (The Blueprint) ---
class PaymentProcessor(ABC):
    
    @abstractmethod
    def validate_payment(self, amount):
        """Abstract method: Every payment gateway must validate payments differently."""
        pass

    @abstractmethod
    def process_transaction(self, amount):
        """Abstract method: Every payment gateway must process transactions differently."""
        pass

    def send_receipt(self, email):
        """Concrete method: Normal method that all child classes inherit automatically."""
        print(f"Receipt successfully sent to {email}")


# --- CHILD CLASS 1: Stripe ---
class StripePayment(PaymentProcessor):
    # Stripe must implement both abstract methods to work
    def validate_payment(self, amount):
        print(f"Stripe: Validating tokenized credit card for ${amount}...")
        return True

    def process_transaction(self, amount):
        print(f"Stripe: Charging ${amount} via Stripe API.")


# --- CHILD CLASS 2: PayPal ---
class PayPalPayment(PaymentProcessor):
    # PayPal must implement both abstract methods to work
    def validate_payment(self, amount):
        print(f"PayPal: Checking user login session and balance for ${amount}...")
        return True

    def process_transaction(self, amount):
        print(f"PayPal: Charging ${amount} via PayPal Wallet.")


# --- TESTING ABSTRACTION ---

# 1. Trying to create the abstract class directly will fail
# processor = PaymentProcessor() 
# TypeError: Can't instantiate abstract class PaymentProcessor with abstract methods...

# 2. Using the concrete implementation (Stripe)
stripe_payment = StripePayment()
stripe_payment.validate_payment(100)      # Output: Stripe: Validating tokenized credit card...
stripe_payment.process_transaction(100)  # Output: Stripe: Charging $100 via Stripe API.
stripe_payment.send_receipt("user@example.com") # Output: Receipt successfully sent...

print("-" * 30)

# 3. Using the concrete implementation (PayPal)
paypal_payment = PayPalPayment()
paypal_payment.validate_payment(50)       # Output: PayPal: Checking user login session...
paypal_payment.process_transaction(50)   # Output: PayPal: Charging $50 via PayPal Wallet.
paypal_payment.send_receipt("admin@example.com") # Output: Receipt successfully sent...
```
