### Polymorphism
The word "polymorphism" means "many forms", and in programming it refers to methods/functions/operators with the same name that can be executed on many objects or classes.

Two types of Polymorphism we can see
1) Function Polymorphism
2) Class Polymorphism

#### Function Polymorphism
An example of a Python function that can be used on different objects is the `len()` function.

```python
x = "Hello World!"
print(len(x)) # It will return the number of characters

mytuple = ("apple", "banana", "cherry")
print(len(mytuple)) # It will return the number of items in the tuple

thisdict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
print(len(thisdict)) # It will return the number of key/value pairs in the dictionary
```

#### Class Polymorphism
Polymorphism is often used in Class methods, where we can have multiple classes with the same method name.

```python
class Phone: # Parent Class
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model
    
    def call(self):
        print("Calling from normal phone")
            

class CameraPhone(Phone): # Child Class
    def __init__(self, brand, model):
        super().__init__(brand, model)

    def call(self):
        print("calling from camera phone")

nokia = Phone('Nokia', '1100')
nokia.call() # It will use call method of parent class

nokia_c101 = CameraPhone('Nokia', 'C101')
nokia_c101.call() # It will use call method of child class
```
