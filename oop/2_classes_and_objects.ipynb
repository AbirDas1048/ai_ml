### Create a class
A Class is like an object constructor, or a "blueprint" for creating objects.

### Create Properties
Properties are variables that belong to a class. They store data for each object created from the class.

There are two types of Properties.

Properties defined outside methods belong to the class itself `class properties` and are shared by all objects.

Properties defined inside `__init__()` belong to each object it's called `instance properties`.

```python
class Phone: # creating a class name is phone

    category = "Electronics" # class properties
```

### Creating objects of a class
### access the class properties of that class by the object

```python
apple = Phone() # creating an object of class phone
print(apple.category) # accessing class property using object
```

### Creating multiple objects
### changing class properties vlaue using object.

```python
apple = Phone() # creating an object of class phone
nokia = Phone() # creating another object of class phone
samsung = Phone() # creating another object of class phone

print(apple.category) # accessing class property using object

apple.category = "Mobile" # changing the value of class property using object
print(apple.category) # accessing class property using object
print(nokia.category) # accessing class property using another object
print(samsung.category) # accessing class property using another object
```

### Delete Objects
We can delete objects by using the `del` keyword

```python
samsung = Phone() # creating object of class phone
print(samsung.category) # accessing class property using object
del samsung # deleting an object of class phone
# print(samsung.category) # accessing class property using object. This will give an error because the object has been deleted and is no longer accessible.
```

### The pass Statement
class definitions cannot be empty, but for some reason have a class definition with no content, put in the pass statement to avoid getting an error.

```python
class Car:
  pass

# With the pass statement, we can create a class without any properties or methods.
# The pass statement is a null operation; nothing happens when it executes.
# It’s useful as a placeholder when a statement is required syntactically but you have nothing to write.
```
