### The `__init__()` Method
All classes have a built-in method called __init__(), which is always executed when the class is being initiated.

The `__init__()` method is used to assign values to object properties, or to perform operations that are necessary when the object is being created.

There are two types of attributes.

Properties defined outside methods belong to the class itself `class properties` and are shared by all objects.

Properties defined inside `__init__()` belong to each object it's called `instance properties`.

```python
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

p1 = Person("Emil", 36)

print(p1.name)
print(p1.age)
```

#### Why Use `__init__()`?
Without the `__init__()` method, we would need to set properties manually for each object.
Using `__init__()` makes it easier to create objects with initial values

```python
# without the __init__() function, we would have to set the properties manually like this:

class Person:
  pass

p1 = Person()
p1.name = "Tobias"
p1.age = 25

print(p1.name)
print(p1.age)
```

#### Default Values in `__init__()`
We can also set default values for parameters in the `__init__()` method:

```python
# Here, we have defined a class named Phone with an __init__() method. 
# the __init__() method takes five parameters: brand, model, price, ram, and storage. 
# The ram and storage parameters have default values of 4 and 64, respectively. 
# This means that if we create a Phone object without specifying the ram and storage values, they will default to 4 and 64.

class Phone:
  def __init__(self, brand, model, price, ram=4, storage=64):
    self.brand = brand
    self.model = model
    self.price = price
    self.ram = ram
    self.storage = storage

# Now, we can create Phone objects with or without specifying the ram and storage values.
apple = Phone("Apple", "iPhone 13", 999, 6, 128)
samsung = Phone("Samsung", "Galaxy S21", 799, 8, 256)
redmi = Phone("Redmi", "Note 10", 199) #  Here, ram and storage will default to 4 and 64, respectively.
honor = Phone("Honor", "50", 499, 8, 128)

print(apple.brand, apple.model, apple.price, apple.ram, apple.storage)
print(samsung.brand, samsung.model, samsung.price, samsung.ram, samsung.storage)
print(redmi.brand, redmi.model, redmi.price, redmi.ram, redmi.storage) 
print(honor.brand, honor.model, honor.price, honor.ram, honor.storage)
```
