### Q. Digits (https://codeforces.com/group/MWSDmqGsZm/contest/219432/problem/Q)

```python
test = int(input())

for i in range(test):

    number = int(input())

    if number == 0:
        print(0)
        continue

    while (number):
        print(number%10, end=" ")
        number //= 10
    print()
```
