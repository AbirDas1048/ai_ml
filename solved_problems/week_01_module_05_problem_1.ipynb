### J. Count Letters (https://codeforces.com/group/MWSDmqGsZm/contest/219856/problem/J)

```python
text = input().lower()
alphabet_count = {}

for letter in text:
  alphabet_count[letter] = alphabet_count.get(letter, 0) + 1

for letter in sorted(alphabet_count.keys()):
    print(f"{letter} : {alphabet_count[letter]}")
```
