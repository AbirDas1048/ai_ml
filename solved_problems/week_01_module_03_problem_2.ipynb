### C. Even, Odd, Positive and Negative (https://codeforces.com/group/MWSDmqGsZm/contest/219432/problem/C)

```python
n = int(input())
numbers = input().split()
even = odd = positive = negative = 0

for i in range(n):
    number = int(numbers[i])
    if number % 2 == 0:
        even += 1
    else:
        odd += 1

    if number > 0:
        positive += 1
    elif number < 0:
        negative += 1
    
print("Even:", even)
print("Odd:", odd)
print("Positive:", positive)
print("Negative:", negative)
```
