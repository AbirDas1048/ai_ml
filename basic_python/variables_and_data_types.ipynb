### Declare variables, types of variables, with print

#### declare variable, assign values and print the value of variables with types 

```python
a = 7
print(a, type(a))
```

```python
a = 10  # integer
b = 3.14  # float
c = "Hello, World!"  # string
is_married = True  # boolean
print(a, type(a))
print(b, type(b))
print(c, type(c))
print(is_married, type(is_married))
```

#### Type Casting

```python
a = 10
b = float(a)  # convert integer to float
c = str(a)  # convert integer to string
print(a, type(a))
print(b, type(b))
print(c, type(c))

d = 3.14
e = int(d)  # convert float to integer
f = str(d)  # convert float to string
print(d, type(d))
print(e, type(e))
print(f, type(f))
```
