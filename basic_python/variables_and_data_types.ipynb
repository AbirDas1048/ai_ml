### Declare variables, types of variables, with print

##### declare variable, assign values and print the value of variables with types 

```python
a = 7
print(a, type(a))
```

```python
a = 10  # integer
b = 3.14  # float
c = "Hello, World!"  # string
is_married = True  # boolean
print(a, type(a))
print(b, type(b))
print(c, type(c))
print(is_married, type(is_married))
```
