### `Tuple`

##### Tuples are used to store multiple items in a single variable.

Tuple is one of 4 built-in data types in Python used to store collections of data, the other 3 are List, Set, and Dictionary, all with different qualities and usage.

A tuple is a collection which is ordered and unchangeable.
Tuples are written with round brackets*.

```python
thistuple1 = ("apple", "banana", "cherry")
print(thistuple1, type(thistuple1))

thistuple2 = "apple", "banana", "cherry"
print(thistuple2, type(thistuple2))
```

###### `Tuple Items`
Tuple items are ordered, unchangeable, and allow duplicate values.

Tuple items are indexed, the first item has index [0], the second item has index [1] etc.

###### `Ordered`
When we say that tuples are ordered, it means that the items have a defined order, and that order will not change.

###### `Unchangeable`
Tuples are unchangeable, meaning that we cannot change, add or remove items after the tuple has been created.

###### `Allow Duplicates`
Since tuples are indexed, they can have items with the same value:

```python
tuple1 = ("apple", "banana", "cherry", "apple", "cherry") # A string tuple with duplicate values
tuple2 = (1, 5, 7, 9, 3) # A numeric tuple
tuple3 = (20.5, 21.5, 19.5, 18.5) # A float tuple
tuple4 = (True, False, False) # A boolean tuple
tuple5 = ("abc", 34, True, 40.5, "male") # A tuple with mixed data types

print(tuple1, type(tuple1))
print(tuple2, type(tuple2))
print(tuple3, type(tuple3))
print(tuple4, type(tuple4))
print(tuple5, type(tuple5))

# Creating a tuple with one item
thistuple1 = ("apple",) # A tuple with one item must have a trailing comma
print(thistuple1, type(thistuple1))

#NOT a tuple
thistuple2 = ("apple")
print(type(thistuple2))

# Create an empty tuple:
thistuple3 = ()
print(type(thistuple3))

# tuple() constructor can also be used to create a tuple
thistuple4 = tuple(("apple", "banana", "cherry")) # note the double round-brackets
print(thistuple4, type(thistuple4))
```

### Tuple Length
To determine how many items a tuple has, use the len() function.

```python
tuple1 = ("apple", "banana", "cherry", "apple", "cherry") # A string tuple with duplicate values
tuple2 = (1, 5, 7) # A numeric tuple
tuple3 = ("abc", 34, True, 40.5, "male") # A tuple with mixed data types

print(f"Tuple 1: {tuple1}, Len: {len(tuple1)}")
print(f"Tuple 2: {tuple2}, Len: {len(tuple2)}")
print(f"Tuple 3: {tuple3}, Len: {len(tuple3)}")
```
