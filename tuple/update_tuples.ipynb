### Change Tuple Values
Once a tuple is created, we cannot change its values. Tuples are unchangeable, or immutable as it also is called.

But there is a workaround. We can convert the tuple into a list, change the list, and convert the list back into a tuple.

```python
x = ("apple", "banana", "cherry")
print(f"the original tuple x = {x}")
y = list(x)
print(f"the converted list y from tuple x = {y}")
y[1] = "kiwi"
print(f"the modified list y = {y}")
x = tuple(y)

print(f"the updated tuple x = {x}")
```

#### Add Items
1. Convert into a list: Just like the workaround for changing a tuple, we can convert it into a list, add item(s), and convert it back into a tuple.

```python
thistuple = ("apple", "banana", "cherry")
print(f"the original tuple thistuple = {thistuple}")
y = list(thistuple)
print(f"the converted list y from tuple thistuple = {y}")
y.append("orange")
print(f"the modified list y after appending 'orange' = {y}")
thistuple = tuple(y)
print(f"the updated tuple thistuple after appending 'orange' = {thistuple}")
```

2. Add tuple to a tuple. We are allowed to add tuples to tuples, so if we want to add one item, (or many), create a new tuple with the item(s), and add it to the existing tuple.

```python
thistuple = ("apple", "banana", "cherry")
y = ("orange",)
thistuple += y

print(thistuple)
```

#### Remove Items
Tuples are unchangeable, so we cannot remove items from it, but we can use the same workaround as we used for changing and adding tuple items.

```python
thistuple = ("apple", "banana", "cherry")
print(f"the original tuple thistuple = {thistuple}")
y = list(thistuple)
print(f"the converted list y from tuple thistuple = {y}")
y.remove("apple")
print(f"the modified list y after removing 'apple' = {y}")
thistuple = tuple(y)
print(f"the updated tuple thistuple after removing 'apple' = {thistuple}")
```

Or we can delete the tuple completely. The `del` keyword can delete the tuple completely

```python
thistuple = ("apple", "banana", "cherry")
del thistuple
print(thistuple) #this will raise an error because the tuple no longer exists
```
