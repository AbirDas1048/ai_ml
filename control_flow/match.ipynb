#### The Python Match Statement
###### Instead of writing many if..else statements, you can use the match statement.

```python
day = 'Monday'

match day:
    case 'Monday':
        print("It's Monday!")
    case 'Tuesday':
        print("It's Tuesday!")
    case 'Wednesday':
        print("It's Wednesday!")
    case 'Thursday':
        print("It's Thursday!")
    case 'Friday':
        print("It's Friday!")
    case 'Saturday':
        print("It's Saturday!")
    case 'Sunday':
        print("It's Sunday!")
    case _:
        print("Invalid day!")
```

```python
match day:
    case 'Monday' | 'Tuesday' | 'Wednesday' | 'Thursday' | 'Friday':
        print("It's a weekday.")
    case 'Saturday' | 'Sunday':
        print("It's a weekend.")
    case _:
        print("Invalid day!")
```
