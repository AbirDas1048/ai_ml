#### The while Loop
###### With the while loop we can execute a set of statements as long as a condition is true.

```python
i = 1
while i < 6:
  print(i)
  i += 1
```

```python
i = 1
while i < 11:
    if i % 2 == 0:
        print(f"{i} is even")
    else:
        print(f"{i} is odd")

    i += 1
```
