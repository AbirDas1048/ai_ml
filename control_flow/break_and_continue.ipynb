#### Python break and continue
##### In Python, break and continue are loop control statements used to alter the normal flow of for and while loops.

• `break` terminates the loop completely and moves control to the next line of code outside the loop.

• `continue` skips the remaining code inside the loop for the current iteration only, immediately jumping to the next iteration.

##### The break Statement
###### With the break statement we can stop the loop even if the while condition is true:

```python
i = 1
while i < 6:
  print(i)
  if i == 3:
    break
  i += 1
```

```python
for i in range(1, 6):
    print(i)
    if i == 3:
        break
```

##### The continue Statement
###### With the continue statement we can stop the current iteration, and continue with the next

```python
i = 0
while i < 6:
  i += 1
  if i == 3:
    continue
  print(i)
```

```python
for i in range(1, 7):
    if i == 3:
        continue
    print(i)
```
