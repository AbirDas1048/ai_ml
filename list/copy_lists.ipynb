#### Copy a List
We cannot copy a list simply by typing list2 = list1, because: list2 will only be a reference to list1, and changes made in list1 will automatically also be made in list2.

```python
thislist1 = ["orange", "mango", "kiwi", "pineapple", "banana"]
thislist2 = thislist1
print(f"Original list: {thislist1}")
print(f"Copied list: {thislist2}")
thislist1[0] = "apple"
print(f"Modified original list: {thislist1}")
print(f"Copied list after original modification: {thislist2}")
thislist2[2] = "grape"
print(f"Modified copied list: {thislist2}")
print(f"Original list after copied modification: {thislist1}")
```

##### Use the copy() method
We can use the built-in List method copy() to copy a list.

```python
thislist = ["orange", "mango", "kiwi", "pineapple", "banana"]
mylist = thislist.copy()
print(f"Original list: {thislist}")
print(f"Copied list: {mylist}")
thislist[0] = "apple"
print(f"Modified original list: {thislist}")
print(f"Copied list after original modification: {mylist}")
mylist[2] = "grape"
print(f"Modified copied list: {mylist}")
print(f"Original list after copied modification: {thislist}")
```

```python
# Another way to make a copy is to use the built-in method list().
thislist = ["orange", "mango", "kiwi", "pineapple", "banana"]
mylist = list(thislist)
print(f"Original list: {thislist}")
print(f"Copied list: {mylist}")
thislist[0] = "apple"
print(f"Modified original list: {thislist}")
print(f"Copied list after original modification: {mylist}")
mylist[2] = "grape"
print(f"Modified copied list: {mylist}")
print(f"Original list after copied modification: {thislist}")
```

```python
# We can also make a copy of a list by using the `: (slice)` operator.
thislist = ["orange", "mango", "kiwi", "pineapple", "banana"]
mylist = thislist[:]
print(f"Original list: {thislist}")
print(f"Copied list: {mylist}")
thislist[0] = "apple"
print(f"Modified original list: {thislist}")
print(f"Copied list after original modification: {mylist}")
mylist[2] = "grape"
print(f"Modified copied list: {mylist}")
print(f"Original list after copied modification: {thislist}")
```
