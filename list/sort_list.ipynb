#### Sort Lists
###### Sort List Alphanumerically
List objects have a sort() method that will sort the list alphanumerically, ascending, by default.

```python
thislist1 = ["orange", "mango", "kiwi", "pineapple", "banana"]
thislist1.sort()
print(f"Sorted list 1: {thislist1}")

thislist2 = [100, 50, 65, 82, 23]
thislist2.sort()
print(f"Sorted list 2: {thislist2}")

# Sort the list descending
thislist1.sort(reverse = True)
print(f"List 1 (descending): {thislist1}")
thislist2.sort(reverse = True)
print(f"List 2 (descending): {thislist2}")
```

##### Reverse Order

The reverse() method reverses the current sorting order of the elements

```python
thislist = ["banana", "Orange", "Kiwi", "cherry"]
thislist.reverse()
print(f"Reversed list: {thislist}")
```

##### Case Insensitive Sort
By default the `sort()` method is case sensitive, resulting in all capital letters being sorted before lower case letters

```python
thislist = ["banana", "Orange", "Kiwi", "cherry"]
thislist.sort()
print(f"List sorted (case-sensitive): {thislist}")

thislist = ["banana", "Orange", "Kiwi", "cherry"]
thislist.sort(key = str.lower)
print(f"List sorted (case-insensitive): {thislist}")
```

#### Customize Sort Function
We can also customize our own function by using the keyword argument `key = function`

```python
def myfunc(n):
    print(f"n = {n}, abs(n - 50) = {abs(n - 50)}")
    return abs(n - 50)

thislist = [100, 50, 65, 82, 23]
thislist.sort(key = myfunc)
print(f"List sorted by distance from 50: {thislist}")
```
