#### Join Two Lists
There are several ways to join, or concatenate, two or more lists in Python.

```python
list1 = ["a", "b", "c"]
list2 = [1, 2, 3]

list3 = list1 + list2
print(f"Joined list: {list3}")

list4 = [1.5, 2.2, 3.9]

for x in list4:
  list3.append(x)

print(f"Final list: {list3}")

list5 = ["x", "y", "z"]

list3.extend(list5)
print(f"Extended list: {list3}")
```
